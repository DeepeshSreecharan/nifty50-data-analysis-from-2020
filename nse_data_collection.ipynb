```python
import pandas as pd
import yfinance as yf
from google.colab import files

# ----------------------------
# 1. Define Nifty50 tickers
# ----------------------------
nifty50_tickers = [
    "RELIANCE.NS", "TCS.NS", "HDFCBANK.NS", "INFY.NS", "ICICIBANK.NS",
    "HINDUNILVR.NS", "SBIN.NS", "KOTAKBANK.NS", "ITC.NS", "LT.NS",
    "HCLTECH.NS", "BAJFINANCE.NS", "AXISBANK.NS", "MARUTI.NS",
    "ASIANPAINT.NS", "BAJAJ-AUTO.NS", "NTPC.NS", "ULTRACEMCO.NS", "SUNPHARMA.NS",
    "WIPRO.NS", "TITAN.NS", "POWERGRID.NS", "DIVISLAB.NS",
    "INDUSINDBK.NS", "M&M.NS", "HINDALCO.NS", "DRREDDY.NS", "BHARTIARTL.NS",
    "GRASIM.NS", "UPL.NS", "EICHERMOT.NS", "SHREECEM.NS", "ADANIGREEN.NS",
    "ADANIPORTS.NS", "TATAMOTORS.NS", "JSWSTEEL.NS", "BAJAJFINSV.NS",
    "TECHM.NS", "TATACONSUM.NS", "COALINDIA.NS", "M&MFIN.NS", "TATAPOWER.NS",
    "IOC.NS", "ONGC.NS", "RECLTD.NS", "INDIGO.NS", "CIPLA.NS"
]

# ----------------------------
# 2. Download all tickers at once
# ----------------------------
print("Downloading all tickers in batch...")
try:
    data = yf.download(
        tickers=nifty50_tickers,
        start="2020-01-01",
        end="2025-01-01",
        group_by='ticker',  # Keeps each ticker as top-level column
        auto_adjust=False,
        progress=True
    )
except Exception as e:
    print("Error downloading data:", e)

# ----------------------------
# 3. Transform to tidy format
# ----------------------------
all_data = []

for ticker in nifty50_tickers:
    if ticker not in data.columns.levels[0]:
        print(f"No data for {ticker}, skipping.")
        continue
    df_ticker = data[ticker].copy()
    df_ticker['Symbol'] = ticker
    df_ticker.reset_index(inplace=True)  # Make Date a column
    all_data.append(df_ticker)

if not all_data:
    raise ValueError("No data was downloaded!")

df = pd.concat(all_data, ignore_index=True)

# Reorder and clean columns
cols = ['Date','Symbol','Open','High','Low','Close','Adj Close','Volume']
df = df[[c for c in cols if c in df.columns]]

# ----------------------------
# 4. Save and download
# ----------------------------
df.to_csv("nifty50_tidy.csv", index=False)
files.download("nifty50_tidy.csv")

print("✅ Tidy dataset created and downloaded as nifty50_tidy.csv")
print(df.head())
```
